```python

def display(value):
    pass

cv2.namedWindow('output')
cv2.createTrackbar('value','output',128,255,display)



while True:
    gray = cv2.imread('ant.jpg',0)
    theshold = cv2.getTrackbarPos('value','output')
    
    _,th = cv2.threshold(gray,theshold,255,cv2.THRESH_BINARY)

    if cv2.waitKey(1) & 0xff==ord('e'):
        break
    cv2.imshow('output',th)


cv2.destroyAllWindows()

```
