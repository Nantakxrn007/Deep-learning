```python
import cv2
```

```python
gray = cv2.imread('gradient.png')

threshold,result = cv2.threshold(gray,120,255,cv2.THRESH_BINARY)
print(threshold) # value threshold 

cv2.imshow('original',gray)
cv2.imshow('binary',result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray = cv2.imread('gradient.png')

threshold,result = cv2.threshold(gray,120,255,cv2.THRESH_BINARY_INV)
print(threshold) # value threshold 

cv2.imshow('original',gray)
cv2.imshow('binary',result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
