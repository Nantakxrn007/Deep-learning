```python
import cv2 
```

```python
cap = cv2.VideoCapture('Walking.mp4')

# check , frame1 = cap.read()
# check , frame2 = cap.read()


while cap.isOpened():
    check , frame1 = cap.read()
    if check == True:
        gray = cv2.cvtColor(frame1,cv2.COLOR_BGR2GRAY)
        _,th = cv2.threshold(gray,215,255,cv2.THRESH_BINARY)


        countors,_ =  cv2.findContours(th,cv2.RETR_TREE,cv2.CHAIN_APPROX_NONE)
        cv2.drawContours(frame1,countors,-1,(0,255,0),2)

        if cv2.waitKey(1) & 0xff == ord('e'):
            break 

        cv2.imshow('output',frame1)
    else:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
cap = cv2.VideoCapture('Walking.mp4')

check , frame1 = cap.read()
check , frame2 = cap.read()


while cap.isOpened():
    if check:
        motiondiff = cv2.absdiff(frame1,frame2)
        gray = cv2.cvtColor(motiondiff,cv2.COLOR_BGR2GRAY)
        _,th = cv2.threshold(gray,15,255,cv2.THRESH_BINARY)
        
        dilation = cv2.dilate(th,None,iterations=3)
        contours,_ =  cv2.findContours(dilation,cv2.RETR_TREE,cv2.CHAIN_APPROX_NONE)
        

        for cnt in contours:
            (x,y,w,h) = cv2.boundingRect(cnt)
            cv2.rectangle(frame1,(x,y),(x+w,y+h),(0,255,0),2)
            if cv2.contourArea(cnt) < 700:
                continue


        cv2.imshow('output',frame1)

        frame1=frame2
        check,frame2 = cap.read()
        if cv2.waitKey(1) & 0xff == ord('e'):
            break 
    else:
        break 

cap.release()
cv2.destroyAllWindows()
```
