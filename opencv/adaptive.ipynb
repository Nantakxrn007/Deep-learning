```python
import cv2
import matplotlib.pyplot as plt
```

```python
gray = cv2.imread('map.jpg',0)

apt_mean = cv2.adaptiveThreshold(gray,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,7,7)
apt_gu = cv2.adaptiveThreshold(gray,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,9,4)


plt.figure(figsize=(20,20))
plt.subplot(121)
plt.imshow(apt_mean,cmap='gray')
plt.subplot(122)
plt.imshow(apt_gu,cmap='gray')
```
