```python
import cv2 
import numpy 
```

```python
img=cv2.imread('ball2d.jpg')
imgresize=cv2.resize(img,(400,400))

lower = numpy.array([5,111,10]) # BGR !!! not RGB 
upper = numpy.array([124,255,133])

mask = cv2.inRange(imgresize,lower,upper)
result = cv2.bitwise_and(imgresize,imgresize,mask=mask)

cv2.imshow('output',imgresize)
cv2.imshow('mask',mask)
cv2.imshow('result',result)
cv2.waitKey(0) 
cv2.destroyAllWindows()
```
