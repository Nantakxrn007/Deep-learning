```python
import cv2

def nothing(x):
    pass

cv2.namedWindow('output')

# Trackbars
cv2.createTrackbar('grid', 'output', 3, 100, nothing)       # Block size (odd only)
cv2.createTrackbar('c', 'output', 0, 100, nothing)          # C value (constant)

while True:
    # โหลดรูป
    gray = cv2.imread('hsv1.jpg', 0)  # โหลด grayscale โดยตรง
    gray = cv2.resize(gray,(720,480))
    # อ่านค่าจาก trackbars
    grid = cv2.getTrackbarPos('grid', 'output')
    c = cv2.getTrackbarPos('c', 'output')

    # ปรับ grid ให้เป็นเลขคี่ ≥ 3
    if grid < 3:
        grid = 3
    if grid % 2 == 0:
        grid += 1

    # ทำ adaptive threshold
    binary = cv2.adaptiveThreshold(
        gray, 255, 
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
        cv2.THRESH_BINARY, 
        grid, c
    )

    cv2.imshow('output', binary)

    if cv2.waitKey(1) & 0xFF == ord('e'):
        break

cv2.destroyAllWindows()
```
