```python
import torch
```

command
- torch.emtry(dim1,dim2,......)

```python
print(torch.empty(1)) # tensor 1 dim 
print(torch.empty(1,2)) # tensor matrix row and column
print(torch.empty(4,2)) # tensor matrix row and column
print(torch.empty(2,4,3)) # mini-batch,row,columns
```

command   
- torch.rand() range 0,1
- torch.zeros() like numpy   
- torch.ones()

```python
print(torch.rand(3,3)) # 0 - 1
print(torch.zeros(3,3))
print(torch.ones(3,3))
```

command
- data type check

```python
x = torch.ones(3,3)
print(x) 
print(x.dtype) # default float32

# we can also give it the following parameter ourselves.
y = torch.ones(3,3 , dtype=torch.int)
print(y)

z = torch.zeros(3,3 , dtype=torch.double)
print(z)
```

command 
- size
- operation

```python
x = torch.tensor([[1,2]])
print(x)
print(x.size())
```

```python
x = torch.tensor([[1,2,3]])
y = torch.tensor([[1,2,3]])
print(x+y) # these the same thing 
print(torch.add(x,y)) # minus sub , multipication mul , division div 
```

command 
- slicing operation 

```python
x = torch.rand(5,3)
display(x)
display(x[:,0]) #all the rows but 1 columns
```

```python
print(x[1,1]) #dim reduct 
print(x[1,1].item()) #actual value, be careful must 1 element only
```

command 
- resize

```python
x = torch.rand(4,4)
print(x.size())
print(x.view(-1).size())
print(x.view(-1)) # break down 
print(x.view(-1,8)) # break down division 2 
#view same reshape
```

command 
- change numpy to tensor 
- change tensor to numpy 

```python
import numpy as np
x = np.ones(5) # numpy 
print(x) #array
y = torch.from_numpy(x)
print(y)
z = torch.tensor(x)
print(z)
```

```python
x = torch.ones(5)
print(x)
y = x.numpy()
print(y)
```

```python
# be careful here 
x = torch.ones(5)
print(x)
y = x.numpy()
print(y)
z = x.add_(1)
print(z) # same memory but GPU not happend 
print(y)
```

```python
if torch.cuda.is_available():
    device = torch.device('cuda')
    print(torch.cuda.get_device_name(0))
    z = torch.tensor([[1]]).to(device=device)
    print(z)
```
