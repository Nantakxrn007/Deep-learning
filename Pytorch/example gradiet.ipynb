```python
import torch

w = torch.tensor(1.0,requires_grad=True)
x = torch.tensor(1.0)
y = torch.tensor(2.0)

y_hat = w * x
loss = (y_hat - y)**2
# dL/dw = 2(wx-y)x
# take value = 2(1-2)1 = -2

# forward 
print(loss)

#backward 
loss.backward()
print(w.grad)

```

```python
import torch


X = torch.tensor([1,2,3,4],dtype=torch.float32)
Y = torch.tensor([2,4,6,8],dtype=torch.float32)
w = torch.tensor(0.0,requires_grad=True,dtype=torch.float32)

#f = wx

def forward(x):
    return w*x

def loss(y_predicted,y):
    return ((y_predicted - y)**2).mean()


#grdient descent 
learning_rate = 0.1
n_iters = 10

for epoch in range(n_iters):
    # forward pass 
    y_pred = forward(X)
    l = loss(y_pred,Y)
    
    # backward pass 
    l.backward()

    #update 
    with torch.no_grad(): #stop add node
        w -= learning_rate*w.grad


    w.grad.zero_()
    print(f'epoch {epoch+1} MSE {l:.4}')
    print(y_pred)
    print(Y)
    print("*"*50)

```
