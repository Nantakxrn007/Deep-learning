- Gradient pytorch 

```python
import torch 
```

```python
# we have say f(x) = 2x^3 
# f'(x) = 6x^2 
# and if let's say x = 2 then 6(2)^2 = 24

x = torch.tensor(2.0 , requires_grad=True)
print(x)

f = 2*x**3
print(f) #forward 2*(2**3) = 16 

f.backward()
print(x.grad)
```

```python
# gradient descent f(x) = 3x^2 
# alpha = 2, initial_value = 4, loop = 2 times


x = torch.tensor(4.0, requires_grad=True)
alpha = 0.1  # learning rate


# p' = p - 0.1(gradient(f(x)))
for i in range(2):  # ทำ 2 รอบ 
    # 1. Forward
    f = 3 * x ** 2
    print(f"Step {i+1} - x: {x.item():.4f}, f(x): {f.item():.4f}")

    # 2. Backward
    f.backward()  # คำนวณ gradient ของ f เทียบกับ x = 6x
    print(f"        df/dx (grad): {x.grad.item():.4f}")

    # 3. Gradient descent update: x = x - alpha * grad
    with torch.no_grad():  # ปิดการ track gradient ตอนอัปเดตค่า
        x -= alpha * x.grad

    # 4. Clear gradient เก่า (สำคัญมาก)
    x.grad.zero_()
```
